```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score , classification_report
import matplotlib.pyplot as plt
import seaborn as sns
from xgboost import XGBClassifier
```

```python
df=pd.read_csv('/content/red_laterite_zone_2020_2025.csv')
df.head()
```

```python
df.shape
```

```python
print(df['District'].unique())
print(df['Target_Crops'].unique())
```

```python
df = df.drop(['Zone','Date','Drought_Stress_Index'], axis=1)
df.columns
```

```python
df.info()
```

```python
le=LabelEncoder()
df['District']=le.fit_transform(df['District'])
df['Target_Crops']=df['Target_Crops'].map({'Pulses_Arhar':0,'Millets_Ragi':1,'Maize':2,'Upland_Rice':3})
print(df['District'].unique())
print(df['Target_Crops'].unique())
```

```python
df.info()
```

```python
X=df.drop('Target_Crops',axis=1)
y=df['Target_Crops']

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=45)
```

```python
rfc_model=RandomForestClassifier()
rfc_model.fit(X_train,y_train)
y_pred=rfc_model.predict(X_test)
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
xgb_model=XGBClassifier()
xgb_model.fit(X_train,y_train)
y_pred=xgb_model.predict(X_test)
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```
